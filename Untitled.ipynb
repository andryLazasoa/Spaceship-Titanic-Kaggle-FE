```python
def missing_values_table(df) :
    # Total missing values
    mis_val = df.isnull().sum()

    # Percentage of missing values
    mis_val_percent = 100*df.isnull().sum() / len(df)

    # Make a table with results
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis = 1)

    # Rename the columns
    mis_val_table_ren_columns = mis_val_table.rename(columns ={0:'Missing Values', 1: '% of Total Values'})

    # Sort the table by percentage of missing descending
    mis_val_table_ren_columns = mis_val_table_ren_columns[mis_val_table_ren_columns.iloc[:,1]!=0].sort_values('% of Total Values', ascending = False).round(1)

    # Print some summary information
    print('Your selected dataframe has ' + str(df.shape[1]) + ' columns.\n'
          'There are ' + str(mis_val_table_ren_columns.shape[0]) + ' columns that have missing values !')

    # Return the dataframe with missing information
    return mis_val_table_ren_columns

missing_values_table(df)
```

```python
def to_strip(df) :
    df['FamilyName'] = df['Name'].str.split(' ').str[1]
    df = df.drop(columns = ['Name'])
    df[['Deck', 'Num', 'Side']] = df['Cabin'].str.split('/', expand = True)
    df['Num'] = pd.to_numeric(df['Num'])
    df = df.drop(columns = ['Cabin'])
    return df
```

```python
df = to_strip(df)
```

```python
def new_cols(df) :
    df['Group'] = df['PassengerId'].str.split('_').str[0]
    df['GroupSize'] = df.groupby('Group')['Group'].transform('count')
    df['IsAlone'] = (df['GroupSize'] == 1).astype(int)
    df['TotalSpent'] = df['RoomService'] + df['FoodCourt'] + df['ShoppingMall'] + df['Spa'] + df['VRDeck']
    return df.head()
```

```python
most_frequent_cols = ['Destination', 'FamilyName', 'VIP', 'HomePlanet',
                      'CryoSleep', 'Deck', 'Num', 'Side']
null_cols = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck', 'TotalSpent']

def impute(df) :
    median_imputer = SimpleImputer(strategy = 'median')
    mf_imputer = SimpleImputer(strategy = 'most_frequent')
    df['Age'] = median_imputer.fit_transform(pd.DataFrame(df['Age']))
    df[null_cols] = df[null_cols].fillna(0)
    df[most_frequent_cols] = mf_imputer.fit_transform(df[most_frequent_cols])
    return df
```

```python
def encode(X):
    X = X.copy()
    X = X.drop(columns = ['PassengerId', 'Group'], errors = 'ignore')
    for c in X.select_dtypes(['object', 'category']).columns:
        X[c] = X[c].astype('category').cat.codes
    return X

def score_dataset(X, y, model=None):
    model = model or GradientBoostingClassifier(random_state=0)
    return cross_val_score(model, encode(X), y, cv=5, scoring='accuracy').mean()

# def score_dataset_light(X, y, model=None):
#     model = model or LGBMClassifier(random_state=0)
#     return cross_val_score(model, encode(X), y, cv=5, scoring='accuracy').mean()
```

```python
def make_mi_score(df) :
    X = df.copy()
    for c in X.select_dtypes('category') :
        X[c], _ = X[c].factorize()
    discrete_features = [pd.api.types.is_integer_dtype(t) for t in X.dtypes]
    mi_score = mutual_info_classif(X, y, discrete_features = discrete_features, random_state = 0)
    mi_score = pd.Series(mi_score, name = 'MI Score', index = X.columns)
    mi_score = mi_score.sort_values(ascending = False)
    return mi_score
```

```python
def plot_mi_scores(scores):
    scores = scores.sort_values(ascending=True)
    width = np.arange(len(scores))
    ticks = list(scores.index)
    plt.barh(width, scores)
    plt.yticks(width, ticks)
    plt.title("Mutual Information Scores")
```

```python
X = df.copy()
y = X.pop('Transported')
X = encode(X)
score = make_mi_score(X)
```

```python
plt.figure(dpi = 80, figsize=(10,8))
plot_mi_scores(score)
```

```python
features = ['FamilyName', 'CryoSleep', 'TotalSpent', 'HomePlanet', 'Destination']
cabin_features = ['Deck', 'Num', 'Side']

def cluster_labels(df, features, n_clusters = 10) :
    X = df.copy()
    X_scaled = X[features]
    X_scaled = (X_scaled - X_scaled.mean(axis =0)) / X_scaled.std(axis =0)
    kmeans = KMeans(n_clusters = n_clusters, n_init = 50, random_state = 0)
    X_new = pd.DataFrame()
    X_new['Cluster'] = kmeans.fit_predict(X_scaled)
    return X_new

def cluster_distance (df, features, n_clusters =10) :
    X = df.copy()
    X_scaled = X[features]
    X_scaled = (X_scaled - X_scaled.mean(axis = 0)) / X_scaled.std(axis =0)
    kmeans = KMeans(n_clusters = n_clusters, n_init = 50, random_state = 0)
    X_trans = kmeans.fit_transform(X_scaled)
    X_trans = pd.DataFrame(
        X_trans, columns=[f'Centroid_{i}' for i in range(X_trans.shape[1])]
    )
    return X_trans
```
